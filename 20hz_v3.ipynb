```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm.notebook import tqdm

# Set style for plots
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette("deep")

# Load the dataset
print("Loading dataset...")
file_path = '/home/understressengineer/programming/SRM_PS1/processed_data/merged_data_20Hz.csv'
df = pd.read_csv(file_path)
print(f"Loaded {len(df)} records.")

# Basic dataset exploration
print("Dataset shape:", df.shape)
print("\nFirst 5 rows:")
display(df.head())

# Identify the Point Machine ID column
point_id_col = 'Point Machine Name'
print(f"\nUsing '{point_id_col}' as point machine identifier with {df[point_id_col].nunique()} unique machines")

# Select relevant features to check for missing values
# These are likely the sensor or measurement columns
relevant_features = ['A Current', 'A Voltage', 'B Current', 'B Voltage']
print(f"\nChecking missing values in features: {relevant_features}")

# Calculate missing values per record
df['missing_count'] = df[relevant_features].isna().sum(axis=1)
df['missing_proportion'] = df['missing_count'] / len(relevant_features)

# Set a threshold for anomaly detection
missing_threshold = 0.25  # Consider as anomaly if 25% or more values are missing
df['anomaly_missing'] = (df['missing_proportion'] >= missing_threshold).astype(int)

# Display results
print(f"\nUsing threshold of {missing_threshold*100}% missing values")
print(f"Identified {df['anomaly_missing'].sum()} anomalous records ({df['anomaly_missing'].mean()*100:.2f}% of data)")

# Calculate anomaly rates per point machine
print("\nCalculating anomaly rates per point machine...")
anomaly_by_machine = df.groupby(point_id_col).agg(
    anomaly_rate=('anomaly_missing', 'mean'),
    missing_avg=('missing_proportion', 'mean'),
    record_count=('anomaly_missing', 'count')
).sort_values('anomaly_rate', ascending=False)

# Display results for machines with sufficient records
min_records = 100  # Minimum records needed for reliable analysis
filtered_anomalies = anomaly_by_machine[anomaly_by_machine['record_count'] >= min_records].copy()

print(f"\nAnomaly rates for point machines with at least {min_records} records (showing top 10):")
display(filtered_anomalies.head(10))

# Visualize top machines by anomaly rate
print("\nCreating anomaly rate visualization...")
plt.figure(figsize=(16, 8))
filtered_anomalies.head(20)['anomaly_rate'].plot(kind='bar')
plt.title(f'Missing Value Anomaly Rates by Point Machine (Top 20, min {min_records} records)')
plt.xlabel('Point Machine Name')
plt.ylabel('Proportion of Anomalies')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()

# Distribution of missing proportions
plt.figure(figsize=(14, 6))
sns.histplot(df['missing_proportion'], bins=50, kde=True)
plt.axvline(x=missing_threshold, color='red', linestyle='--', 
           label=f'Threshold ({missing_threshold*100}%)')
plt.title('Distribution of Missing Value Proportions')
plt.xlabel('Proportion of Missing Values')
plt.ylabel('Frequency')
plt.legend()
plt.tight_layout()
plt.show()

# Analyze the top anomalous machines
top_anomalous = filtered_anomalies.head(5).index.tolist()
print(f"\n===== Detailed Analysis of Top 5 Anomalous Point Machines =====")
print(f"Top 5 anomalous machines: {top_anomalous}")

for machine in tqdm(top_anomalous, desc="Analyzing top anomalous machines"):
    print(f"\n--- Analysis for {machine} ---")
    machine_data = df[df[point_id_col] == machine].copy()
    
    print(f"Total records: {len(machine_data)}")
    print(f"Anomalous records: {machine_data['anomaly_missing'].sum()} ({machine_data['anomaly_missing'].mean()*100:.2f}%)")
    print(f"Average missing proportion: {machine_data['missing_proportion'].mean()*100:.2f}%")
    
    # Missing value pattern in the features
    missing_pattern = machine_data[relevant_features].isna().mean().sort_values(ascending=False)
    print("\nMissing value pattern by feature:")
    display(missing_pattern)
    
    # Visualize missing value patterns
    plt.figure(figsize=(12, 6))
    sns.heatmap(machine_data[relevant_features].isna().transpose(), 
               cmap='viridis', cbar_kws={'label': 'Missing'}, 
               yticklabels=relevant_features)
    plt.title(f'Missing Value Pattern for {machine}')
    plt.xlabel('Record Index')
    plt.tight_layout()
    plt.show()
```

```python
# Extended Point Machine Analysis
print("\n===== Extended Point Machine Analysis =====")

# Create a summary of point machines by site
print("Analyzing point machine distribution by site...")
site_machine_counts = df.groupby('Site Name')[point_id_col].nunique().sort_values(ascending=False)
print("\nNumber of point machines per site:")
display(site_machine_counts)

# Create a site-level anomaly analysis
print("\nCalculating site-level anomaly rates...")
site_anomaly = df.groupby('Site Name').agg(
    anomaly_rate=('anomaly_missing', 'mean'),
    missing_avg=('missing_proportion', 'mean'),
    machine_count=(point_id_col, lambda x: x.nunique()),
    record_count=('anomaly_missing', 'count')
).sort_values('anomaly_rate', ascending=False)

# Display site-level results
print("\nAnomaly rates by site:")
display(site_anomaly)

# Visualize site-level anomaly rates
plt.figure(figsize=(14, 7))
ax = site_anomaly['anomaly_rate'].plot(kind='bar')
plt.title('Missing Value Anomaly Rates by Site')
plt.xlabel('Site Name')
plt.ylabel('Proportion of Anomalies')
plt.xticks(rotation=90)

# Add text labels with machine counts
for i, v in enumerate(site_anomaly['anomaly_rate']):
    machines = site_anomaly['machine_count'].iloc[i]
    ax.text(i, v + 0.01, f"{machines} machines", ha='center', fontsize=8)

plt.tight_layout()
plt.show()

# Analyze relationship between direction and anomalies
if 'Direction' in df.columns:
    print("\nAnalyzing relationship between direction and anomalies...")
    direction_anomaly = pd.crosstab(
        df['Direction'], 
        df['anomaly_missing'],
        normalize='index'
    ) * 100
    
    print("\nAnomaly rate by direction (%):")
    display(direction_anomaly)
    
    plt.figure(figsize=(10, 6))
    direction_anomaly[1].sort_values(ascending=False).plot(kind='bar')
    plt.title('Anomaly Rate by Direction')
    plt.xlabel('Direction')
    plt.ylabel('Percentage of Anomalies')
    plt.tight_layout()
    plt.show()

# Check if Type columns exist and analyze relationship with anomalies
type_columns = [col for col in df.columns if col.startswith('Type')]
if type_columns:
    print("\nAnalyzing relationship between machine types and anomalies...")
    
    for type_col in type_columns:
        if df[type_col].nunique() <= 10:  # Only analyze if reasonable number of types
            type_anomaly = pd.crosstab(
                df[type_col], 
                df['anomaly_missing'],
                normalize='index'
            ) * 100
            
            print(f"\nAnomaly rate by {type_col} (%):")
            display(type_anomaly)
            
            plt.figure(figsize=(10, 6))
            type_anomaly[1].sort_values(ascending=False).plot(kind='bar')
            plt.title(f'Anomaly Rate by {type_col}')
            plt.xlabel(type_col)
            plt.ylabel('Percentage of Anomalies')
            plt.tight_layout()
            plt.show()

# Create a timeline view for anomalies if timestamp column exists
timestamp_cols = [col for col in df.columns if 'time' in col.lower() or 'date' in col.lower()]
if timestamp_cols:
    time_col = timestamp_cols[0]  # Use the first time-related column
    print(f"\nAnalyzing anomaly occurrence over time using {time_col}...")
    
    # Convert to datetime if not already
    df[time_col] = pd.to_datetime(df[time_col], errors='coerce')
    
    # Group by date and calculate anomaly rate
    if not pd.isna(df[time_col]).all():
        df['date'] = df[time_col].dt.date
        timeline = df.groupby('date').agg(
            anomaly_rate=('anomaly_missing', 'mean'),
            record_count=('anomaly_missing', 'count')
        )
        
        # Plot timeline
        plt.figure(figsize=(16, 6))
        timeline['anomaly_rate'].plot(kind='line')
        plt.title('Anomaly Rate Over Time')
        plt.xlabel('Date')
        plt.ylabel('Proportion of Anomalies')
        plt.grid(True, linestyle='--', alpha=0.7)
        plt.tight_layout()
        plt.show()

# Compare top anomalous machines vs normal machines
print("\n===== Comparative Analysis: Anomalous vs Normal Machines =====")

# Get bottom 5 anomalous machines (with sufficient records)
normal_machines = filtered_anomalies.tail(5).index.tolist()
print(f"Comparing top 5 anomalous machines with 5 normal machines:")
print(f"Top 5 anomalous: {top_anomalous}")
print(f"5 normal machines: {normal_machines}")

# Create comparison table
comparison_data = []
for machine in top_anomalous + normal_machines:
    machine_data = df[df[point_id_col] == machine]
    
    comparison_data.append({
        'Point Machine': machine,
        'Record Count': len(machine_data),
        'Anomaly Rate': machine_data['anomaly_missing'].mean(),
        'Avg Missing %': machine_data['missing_proportion'].mean() * 100,
        'Type': machine_data['Type of A'].iloc[0] if 'Type of A' in df.columns else 'Unknown',
        'Site': machine_data['Site Name'].iloc[0] if 'Site Name' in df.columns else 'Unknown',
    })

# Add feature statistics if available
for feature in relevant_features:
    for i, machine in enumerate(top_anomalous + normal_machines):
        machine_data = df[df[point_id_col] == machine]
        non_missing = machine_data[feature].dropna()
        
        if len(non_missing) > 0:
            comparison_data[i][f'{feature} Mean'] = non_missing.mean()
            comparison_data[i][f'{feature} Std'] = non_missing.std()
        else:
            comparison_data[i][f'{feature} Mean'] = np.nan
            comparison_data[i][f'{feature} Std'] = np.nan

# Convert to DataFrame and display
comparison_df = pd.DataFrame(comparison_data)
print("\nComparative statistics:")
display(comparison_df)

# Create a visualization to compare machines
plt.figure(figsize=(14, 7))
sns.barplot(x='Point Machine', y='Avg Missing %', hue='Type', data=comparison_df)
plt.title('Missing Value Percentage by Machine Type')
plt.xlabel('Point Machine')
plt.ylabel('Average Missing Value %')
plt.xticks(rotation=90)
plt.grid(True, linestyle='--', alpha=0.3)
plt.tight_layout()
plt.show()

# Feature distributions for anomalous vs normal
print("\nComparing feature distributions between anomalous and normal records...")

# Plot feature distributions by anomaly status
if len(relevant_features) > 0:
    fig, axes = plt.subplots(len(relevant_features), 1, figsize=(12, 4*len(relevant_features)))
    if len(relevant_features) == 1:
        axes = [axes]  # Make iterable for single feature case
    
    for i, feature in enumerate(relevant_features):
        # Get non-missing values for this feature
        feature_data = df[[feature, 'anomaly_missing']].dropna()
        
        if len(feature_data) > 0:
            sns.boxplot(x='anomaly_missing', y=feature, data=feature_data, ax=axes[i])
            axes[i].set_title(f'{feature} Distribution by Anomaly Status')
            axes[i].set_xlabel('Is Anomaly')
            axes[i].set_ylabel(feature)
    
    plt.tight_layout()
    plt.show()
```
