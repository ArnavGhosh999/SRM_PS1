```python
import os
import glob
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from collections import defaultdict
import seaborn as sns

# Set up better visualization
plt.style.use('seaborn')
sns.set_context("notebook", font_scale=1.2)

# Directory containing CSV files
data_dir = "/home/understressengineer/programming/SRM_PS1/FEB 2025 Prototype data"

# Get list of CSV files
csv_files = glob.glob(os.path.join(data_dir, "*.csv"))
if not csv_files:
    print("No CSV files found in:", data_dir)
else:
    print(f"Found {len(csv_files)} CSV files")

# Read the CSV files
data = {}
for file_path in csv_files:
    file_name = os.path.basename(file_path)
    try:
        # Skip lines starting with '//' (comments)
        df = pd.read_csv(file_path, comment='/')
        data[file_name] = df
        print(f"Loaded {file_name} with {len(df)} rows and {len(df.columns)} columns")
    except Exception as e:
        print(f"Error reading {file_name}: {e}")

# Check if 'Polling of A' and 'Polling of B' columns exist in the data
polling_columns = []
if data:
    sample_df = next(iter(data.values()))
    if 'Polling of A' in sample_df.columns:
        polling_columns.append('Polling of A')
    if 'Polling of B' in sample_df.columns:
        polling_columns.append('Polling of B')
    
    if not polling_columns:
        print("Warning: No polling columns found. Available columns:", sample_df.columns.tolist())
else:
    print("No data was loaded")

# Group datasets by polling rates
polling_groups = defaultdict(list)
for file_name, df in data.items():
    if not polling_columns:
        print(f"Cannot group {file_name} - no polling columns identified")
        continue
    
    # Get unique polling rates for this file
    polling_rates = []
    for col in polling_columns:
        if col in df.columns:
            unique_rates = df[col].unique()
            polling_rates.extend([f"{col}:{rate}" for rate in unique_rates])
    
    # If we found polling rates, use them as grouping keys
    if polling_rates:
        polling_key = '-'.join(sorted(str(rate) for rate in polling_rates))
        polling_groups[polling_key].append(file_name)
    else:
        # If no polling rates found, put in "unknown" group
        polling_groups["unknown"].append(file_name)

# Display the groups
print("\nDataset groups by polling rates:")
for polling_rate, file_list in polling_groups.items():
    print(f"\nPolling rate: {polling_rate}")
    print(f"Number of files: {len(file_list)}")
    print("Files:", ", ".join(file_list))

# Create combined dataframes for each polling rate group
combined_data = {}
for polling_rate, file_list in polling_groups.items():
    group_dfs = [data[file_name] for file_name in file_list]
    if group_dfs:
        # Get common columns across all dataframes in this group
        common_cols = set.intersection(*[set(df.columns) for df in group_dfs])
        # Get only numeric columns for clustering
        numeric_cols = []
        for col in common_cols:
            if all(pd.api.types.is_numeric_dtype(df[col]) for df in group_dfs):
                numeric_cols.append(col)
        
        if numeric_cols:
            # Concatenate all dataframes, keeping track of source file
            for i, (file_name, df) in enumerate(zip(file_list, group_dfs)):
                df = df[numeric_cols].copy()
                df['source_file'] = file_name
                df['group_id'] = i
                if i == 0:
                    combined_df = df
                else:
                    combined_df = pd.concat([combined_df, df], ignore_index=True)
            
            combined_data[polling_rate] = combined_df
            print(f"Combined data for {polling_rate}: {len(combined_df)} rows, {len(numeric_cols)} numeric features")
        else:
            print(f"No common numeric columns found for group {polling_rate}")
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, DBSCAN, AgglomerativeClustering
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score, calinski_harabasz_score
import warnings
warnings.filterwarnings('ignore')

# Function to perform clustering and visualization
def cluster_and_visualize(data_df, group_name, numeric_cols, n_clusters=3):
    """
    Train clustering models and visualize results
    
    Args:
        data_df: DataFrame containing the data
        group_name: Name of the polling rate group
        numeric_cols: List of numeric columns to use for clustering
        n_clusters: Number of clusters for K-means and Agglomerative 
    """
    # Get feature matrix X
    X = data_df[numeric_cols].copy()
    
    # Handle missing values
    X = X.fillna(X.mean())
    
    # Standardize features
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)
    
    # Apply PCA for visualization
    pca = PCA(n_components=2)
    X_pca = pca.fit_transform(X_scaled)
    
    # Create figure for plotting
    fig, axes = plt.subplots(1, 3, figsize=(18, 5))
    
    # 1. K-means clustering
    kmeans = KMeans(n_clusters=n_clusters, random_state=42, n_init=10)
    kmeans_labels = kmeans.fit_predict(X_scaled)
    
    # 2. DBSCAN clustering
    dbscan = DBSCAN(eps=0.5, min_samples=5)
    dbscan_labels = dbscan.fit_predict(X_scaled)
    
    # 3. Hierarchical clustering
    hierarchical = AgglomerativeClustering(n_clusters=n_clusters)
    hierarchical_labels = hierarchical.fit_predict(X_scaled)
    
    # Store all clustering results
    labels = [kmeans_labels, dbscan_labels, hierarchical_labels]
    model_names = ['K-means', 'DBSCAN', 'Hierarchical']
    
    # Plot results
    for i, (model_label, model_name) in enumerate(zip(labels, model_names)):
        # Plot clusters
        scatter = axes[i].scatter(X_pca[:, 0], X_pca[:, 1], c=model_label, cmap='viridis', alpha=0.7, s=50)
        axes[i].set_title(f'{model_name} Clustering\n{group_name}')
        axes[i].set_xlabel(f'PCA 1 ({pca.explained_variance_ratio_[0]:.2%} variance)')
        axes[i].set_ylabel(f'PCA 2 ({pca.explained_variance_ratio_[1]:.2%} variance)')
        
        # Add colorbar
        cbar = plt.colorbar(scatter, ax=axes[i])
        cbar.set_label('Cluster')

    plt.tight_layout()
    plt.show()
    
    # Evaluate clustering performance
    print(f"\nClustering Evaluation for {group_name}:")
    for i, (model_label, model_name) in enumerate(zip(labels, model_names)):
        # Skip DBSCAN evaluation if it only found noise points (-1)
        if model_name == 'DBSCAN' and len(set(model_label)) <= 1:
            print(f"{model_name}: Found only noise points, skipping evaluation")
            continue
            
        # Skip evaluation for models with only one cluster
        if len(set(model_label)) <= 1:
            print(f"{model_name}: Only one cluster found, skipping evaluation")
            continue
            
        try:
            sil_score = silhouette_score(X_scaled, model_label)
            ch_score = calinski_harabasz_score(X_scaled, model_label)
            print(f"{model_name}: Silhouette Score = {sil_score:.3f}, Calinski-Harabasz Score = {ch_score:.1f}")
        except Exception as e:
            print(f"{model_name}: Evaluation error - {str(e)}")
    
    # Return a dictionary with model results
    return {
        'kmeans': kmeans_labels,
        'dbscan': dbscan_labels,
        'hierarchical': hierarchical_labels,
        'X_pca': X_pca,
        'X_scaled': X_scaled
    }

# Apply clustering to each polling rate group
clustering_results = {}
for group_name, group_df in combined_data.items():
    print(f"\n{'-'*80}")
    print(f"Clustering for group: {group_name}")
    
    # Get numeric columns (excluding metadata columns)
    feature_cols = [col for col in group_df.columns 
                   if col not in ['source_file', 'group_id'] 
                   and pd.api.types.is_numeric_dtype(group_df[col])]
    
    # Determine optimal number of clusters (using a simple heuristic)
    # For more advanced methods, you could implement elbow method or silhouette analysis
    n_samples = len(group_df)
    n_clusters = min(3, max(2, int(np.sqrt(n_samples/10))))
    print(f"Using {n_clusters} clusters based on dataset size of {n_samples} samples")
    
    # Apply clustering
    if len(feature_cols) > 0:
        result = cluster_and_visualize(
            data_df=group_df,
            group_name=group_name,
            numeric_cols=feature_cols,
            n_clusters=n_clusters
        )
        clustering_results[group_name] = result
        
        # Optional: analyze how clusters relate to original source files
        if 'source_file' in group_df.columns:
            kmeans_labels = result['kmeans']
            source_distribution = pd.crosstab(
                group_df['source_file'], 
                kmeans_labels,
                normalize='index'
            )
            print("\nDistribution of source files across K-means clusters:")
            print(source_distribution)
    else:
        print(f"No numeric features found for group {group_name}")
```
